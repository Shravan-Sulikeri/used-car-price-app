```python
# Locate and (re)load the module from ./model or ../model
import sys, importlib
from pathlib import Path

candidates = [Path.cwd() / "model", Path.cwd().parent / "model"]
for p in candidates:
    if (p / "standardize_makes.py").exists():
        sys.path.insert(0, str(p))
        break
else:
    raise FileNotFoundError("standardize_makes.py not found in ./model or ../model")

# Clear any stale module, then import
sys.modules.pop("standardize_makes", None)
sm = importlib.import_module("standardize_makes")
canonicalize_make = sm.canonicalize_make

# quick sanity checks
print("bmw..  ->", canonicalize_make("bmw.."))                 # BMW
print("mercedes benz s class ->", canonicalize_make("mercedes benz s class"))  # Mercedes-Benz
print("vw     ->", canonicalize_make("vw"))                    # Volkswagen
```

```python
import pandas as pd
df = pd.read_csv("/workspaces/used-car-price-app/data/clean_used_cars.csv", low_memory=False)

makes_counts = (df["make"]
    .astype(str)
    .value_counts(dropna=False)
    .rename_axis("make")
    .reset_index(name="count"))

len(makes_counts), makes_counts.head(20)  # shows how many unique + top 20
```

```python
all_makes = sorted(df["make"].dropna().astype(str).unique())
len(all_makes), all_makes[:30]  # preview first 30
```

```python
# Save into the repo-level data folder
from pathlib import Path
out_path = Path("..") / "data" / "makes_unique_counts.csv"
out_path.parent.mkdir(parents=True, exist_ok=True)
makes_counts.sort_values("make").to_csv(out_path, index=False)
print("Wrote:", out_path)
```
